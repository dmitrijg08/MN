### Завдання 25
Дано два вектори:

```text
a = [1, 2]
b = [2, 1]
```

Обчисліть косинус кута між ними за формулою через скалярний добуток і норми векторів.

Додатково знайдіть сам кут у градусах.

````python
# Код студента
a = np.array([1, 2])
b = np.array([2, 1])

# Обчислення скалярного добутку та норм
dot_product = np.dot(a, b)
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)

# Обчислення косинуса кута
cos_theta = dot_product / (norm_a * norm_b)

# Знаходження кута в радіанах за допомогою арккосинуса
angle_rad = np.arccos(cos_theta)

# Переведення радіанів у градуси
angle_deg = np.degrees(angle_rad)

print("Косинус кута:", cos_theta)
print("Кут у градусах:", angle_deg)
````

### Завдання 27
Дано матриці:

```text
A = [[1, 2],
     [3, 4]]

B = [[2, 0],
     [1, 2]]
```

Обчисліть:
1. поелементний добуток `A * B`;
2. матричний добуток `A @ B`.

Поясніть у коментарі до коду, чим відрізняються ці операції.

````python
# Код студента
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [2, 0],
    [1, 2]
])

# 1. Поелементний добуток (A * B або матриця Адамара)
# При цій операції множаться елементи, що стоять на однакових позиціях.
# Наприклад: новий елемент [0,0] = A[0,0] * B[0,0] (1 * 2 = 2).
elementwise_product = A * B

# 2. Матричний добуток (A @ B або np.dot(A, B))
# Це класичне множення матриць з лінійної алгебри за правилом "рядок на стовпець".
# Кожен елемент результату — це сума добутків елементів рядка першої матриці
# на відповідні елементи стовпця другої матриці.
matrix_product = A @ B

print("Поелементний добуток (A * B):\n", elementwise_product)
print("\nМатричний добуток (A @ B):\n", matrix_product)
````

Перевірте результат, помноживши початкову матрицю на обернену.

````python
# Код студента
A = np.array([
    [4, 7],
    [2, 6]
])

# Обчислення визначника
determinant = np.linalg.det(A)

# Обчислення оберненої матриці
inverse_A = np.linalg.inv(A)

# Перевірка (множення оригінальної матриці на обернену)
verification = A @ inverse_A

print("Визначник:", determinant)
print("\nОбернена матриця:\n", inverse_A)

# Через особливості обчислень з плаваючою комою використовуємо np.round
# для коректного відображення нулів в одиничній матриці
print("\nРезультат множення A на обернену (одинична матриця):\n", np.round(verification))
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
A = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

# Обчислення середнього значення (μ) для кожного стовпця
mu = np.mean(A, axis=0)

# Обчислення стандартного відхилення (σ) для кожного стовпця
sigma = np.std(A, axis=0)

# Застосування формули нормалізації (завдяки broadcasting вектори
# mu та sigma автоматично застосовуються до кожного рядка матриці)
Z = (A - mu) / sigma

print("Середні значення стовпців (μ):", mu)
print("Стандартні відхилення стовпців (σ):", sigma)
print("\nНормалізована матриця (Z-оцінка):\n", Z)
````
