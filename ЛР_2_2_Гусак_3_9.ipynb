```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Завантаження та базовий огляд
df = pd.read_csv('titanic.csv')
display(df.head())
print("Пропущені значення:\n", df.isnull().sum())
display(df.describe())

# Візуалізація
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.countplot(data=df, x='Survived', hue='Sex', palette='Set1')
plt.title('Виживання за статтю')

plt.subplot(1, 2, 2)
sns.countplot(data=df, x='Survived', hue='Pclass', palette='Set2')
plt.title('Виживання за класом')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Завантаження та базовий огляд
df = pd.read_csv('bestsellers with categories.csv')
display(df.head())
print("Пропущені значення:\n", df.isnull().sum())
display(df.describe())

# Візуалізація
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
genre_counts = df['Genre'].value_counts()
plt.pie(genre_counts, labels=genre_counts.index, autopct='%1.1f%%', colors=['#ff9999','#66b3ff'])
plt.title('Розподіл за жанрами')

plt.subplot(1, 3, 2)
top_authors = df['Author'].value_counts().head(10)
sns.barplot(x=top_authors.values, y=top_authors.index, palette='viridis')
plt.title('Топ-10 авторів')

plt.subplot(1, 3, 3)
price_trend = df.groupby('Year')['Price'].mean()
sns.lineplot(x=price_trend.index, y=price_trend.values, marker='o', color='green')
plt.title('Середня ціна за роками')

plt.tight_layout()
plt.show()
```
